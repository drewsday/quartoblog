I read [this interesting piece in the New York Times](https://www.nytimes.com/2025/06/16/opinion/sports-nba-mlb-leagues-streaming-fan.html?unlocked_article_code=1.P08.p6cX.CVRgXU2OWQs5&smid=url-share) recently about the cost of paying for all the streaming platforms necessary to follow sports in the year 2025. 

This article really resonated with me, as I find it more and more difficult to pay attention to the teams that I've followed my whole life.  I end up wondering how it will be possible for our son to develop into a fan of any of these teams when we can barely find any sports to play in the house. I'd love to see more Cubs games, but we don't have cable anymore, and it seems like an impossible task to figure out where each game is streaming. In the end, I usually just check box scores on my phone and see random highlights on Facebook.  This is not a recipe for long-term or generational devotion to a team.
