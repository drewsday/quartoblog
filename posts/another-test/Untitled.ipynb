```python
import svgwrite

# Create a new SVG drawing
dwg = svgwrite.Drawing('christmas_fireplace_ornament.svg', profile='tiny', size=("200px", "250px"))

# Draw the hanging loop
dwg.add(dwg.circle(center=("100", "20"), r="10", fill="silver", stroke="black"))

# Draw the ornament base (a rounded rectangle)
dwg.add(dwg.rect(insert=("25", "40"), size=("150", "190"), rx="15", ry="15", fill="#f5f5dc", stroke="black"))

# Draw the fireplace
dwg.add(dwg.rect(insert=("50", "100"), size=("100", "80"), fill="#8b4513", stroke="black"))

# Draw the fire
dwg.add(dwg.polygon(points=[("100", "160"), ("90", "180"), ("100", "170"), ("110", "180")], fill="orange", stroke="red"))

# Draw the mantle
dwg.add(dwg.rect(insert=("45", "90"), size=("110", "10"), fill="#a0522d", stroke="black"))

# Draw stockings
stocking_positions = [60, 95, 130]
for x in stocking_positions:
    dwg.add(dwg.rect(insert=(f"{x}px", "60"), size=("15", "25"), fill="red", stroke="black"))
    dwg.add(dwg.rect(insert=(f"{x}px", "60"), size=("15", "5"), fill="white", stroke="black"))

# Save the SVG file
dwg.save()

print("SVG file 'christmas_fireplace_ornament.svg' has been created and saved.")

```
